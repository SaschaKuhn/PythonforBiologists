### Part2: different data structures

```python
import numpy as np
```

```python
# 1. LIST - Python's basic collection
temperatures = [20, 22, 19, 25, 23, 21, 24]
print("LIST of temperatures:")
print(temperatures)
print(f"Type: {type(temperatures)}")
print(f"Length: {len(temperatures)}")
print()
```

```python
# 2. ARRAY - NumPy's numeric array
temps_array = np.array(temperatures)
print("NUMPY ARRAY of temperatures:")
print(temps_array)
print(f"Type: {type(temps_array)}")
print(f"Shape: {temps_array.shape}")
print(f"Mean: {temps_array.mean()}")
print()
```

```python
# 3. DICTIONARY - Key-value pairs
weather_data = {
    'Monday': 20,
    'Tuesday': 22,
    'Wednesday': 19,
    'Thursday': 25,
    'Friday': 23,
    'Saturday': 21,
    'Sunday': 24
}
print("DICTIONARY of weather data:")
print(weather_data)
print(f"Type: {type(weather_data)}")
print(f"Keys: {list(weather_data.keys())}")
print(f"Thursday's temp: {weather_data['Thursday']}")
print()
```

```python
# 4. PANDAS DATAFRAME - Table with rows and columns
df = pd.DataFrame({
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'Temperature': temperatures,
    'Humidity': [65, 70, 68, 60, 62, 66, 64]
})
print("PANDAS DATAFRAME:")
print(df)
print(f"Type: {type(df)}")
print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
```
