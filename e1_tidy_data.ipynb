# Exercise 1: Analysing 96-well data  

96-well plates are a classic test format in biological experiments. They are often used for experiments carried out in plate readers. Unfortunately, most plate reader software saves the data in an untidy format, the same 12x8 matrix layout as the well plate with numerical column and alphabetical row indices; and mostly as an excel table.  
In the `data` folder you can find the `BCA.xlsx` file. Try to load the individual spread sheets (there should be three) with pd.read_excel (you might have to conda install something, pay attention to potential error messages; they might tell you what to do).  
Try to use the pd.melt function to turn the ugly pplate layout into a tidy data frame.
